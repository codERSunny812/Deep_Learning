```python
import numpy as np
```

```python
# input feature 

input = np.array([5.0,0.7,9.0,9.7])
```

```python
input
```

```python
#weight
weights = np.array([0.8,0.6,0.5,0.1])
```

```python
weights
```

```python
#bais 
b = -1.0
```

```python
#weighted sum 
z = np.dot(input,weights)+b
```

```python
z
```

```python
#activation function 
a = np.maximum(0,z)
```

```python
a
```

```python
print("Weighted sum:", z)
print("Neuron output:", a)
```
